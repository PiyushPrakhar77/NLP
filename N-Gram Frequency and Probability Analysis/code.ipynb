```python
import nltk, re
from nltk.corpus import brown
from nltk import FreqDist
from collections import Counter

nltk.download('brown')

news_text = brown.words(categories='news')
news_sentences = brown.sents(categories='news')

tokens = [re.sub(r'[^a-z]', '', w.lower()) for w in news_text]
tokens = [w for w in tokens if w]

def ngrams(tokens, n):
    return [tuple(tokens[i:i+n]) for i in range(len(tokens)-n+1)]

unigrams = ngrams(tokens, 1)
bigrams = ngrams(tokens, 2)
trigrams = ngrams(tokens, 3)

bigram_counts = Counter(bigrams)
unigram_counts = Counter(tokens)
trigram_counts = Counter(trigrams)
```

```python
print("Total sentences:", len(news_sentences))
print("Total words:", len(tokens))
print("Vocabulary size:", len(set(tokens)))
```

```python
fdist_bigrams = FreqDist(bigrams)
fdist_trigrams = FreqDist(trigrams)

print("Top 20 Bigrams:", fdist_bigrams.most_common(20))
print("Top 20 Trigrams:", fdist_trigrams.most_common(20))
```

```python
def bigram_prob(w1, w2):
    return bigram_counts[(w1,w2)] / unigram_counts[w1] if unigram_counts[w1]>0 else 0

def trigram_prob(w1,w2,w3):
    return trigram_counts[(w1,w2,w3)] / bigram_counts[(w1,w2)] if bigram_counts[(w1,w2)]>0 else 0

print("P('the'|'in') =", bigram_prob('in','the'))
print("P('president'|'the','new') =", trigram_prob('the','new','president'))
```

```python
def sentence_prob_bigram(sentence):
    words = [re.sub(r'[^a-z]', '', w.lower()) for w in sentence.split()]
    words = [w for w in words if w]
    prob = 1
    for i in range(len(words)-1):
        prob *= bigram_prob(words[i], words[i+1])
    return prob

def sentence_prob_trigram(sentence):
    words = [re.sub(r'[^a-z]', '', w.lower()) for w in sentence.split()]
    words = [w for w in words if w]
    prob = 1
    for i in range(len(words)-2):
        prob *= trigram_prob(words[i], words[i+1], words[i+2])
    return prob

s = "the president of the company"
print("Bigram probability:", sentence_prob_bigram(s))
print("Trigram probability:", sentence_prob_trigram(s))
```
