```python
import re
```

```python
def extract_domain(text):
  pattern = r"https?://(?:www\.)?([a-zA-Z0-9.-]+\.[a-zA-Z]{2,})"
  return re.findall(pattern, text)

sentence = "Visit our portal at https://www.iitbhu.ac.in for details."
domains = extract_domain(sentence)
print(domains)
```

```python
def extract_email(text):
 pattern = r"\b[A-Za-z0-9._%+-]+@(?:gmail\.com|yahoo\.com)\b"
 return re.findall(pattern, text)

sentence = "Send updates to abc123@yahoo.com and x.y.z@gmail.com"
emails = extract_email(sentence)
print(emails)
```

```python
def extract_hastag(text):
  pattern = r"#\w*(?:\d|_)\w*"
  return re.findall(pattern, text)

sentence = "Today's highlights: #IndvsPak_2025 #Go_India #Cricket123"
hashtags = extract_hastag(sentence)
print(hashtags)
```

```python
def extract_at_a_rate(text):
  pattern = r"@[A-Za-z0-9_]+"
  return re.findall(pattern,text)

sentence = "@dr_singh! you were amazing @Minister@India"
at = extract_at_a_rate(sentence)
print(at)
```

```python
def extract_decimal(text):
  pattern = r"\b\d+\.\d+\b"
  return re.findall(pattern, text)

sentence = "The price increased from 45.60 to 99.99 in 3 days"
decimal = extract_decimal(sentence)
print(decimal)
```

```python
def extract_special_character(text):
  pattern = r"[^a-zA-Z0-9\s]"
  return re.findall(pattern, text)

sentence = "Whoa!! COVID-19 has changed the world @2020 #History";
special_char = extract_special_character(sentence)
print(special_char)
```

```python
def validate_number(text):
  pattern = r"^[A-Z]{2}\d{2}[A-Z]{2}\d{4}$"
  if re.match(pattern, text):
    return True
  return False

sentence = "UP32GH1234"
print(validate_number(sentence))
```

```python
def normalise_text(text):
  return re.sub(r"(.)\1+", r"\1", text)

sentence = "Ammmaaazing wooorrrkkk!!!"
print(normalise_text(sentence))
```

```python
def extract_number(text):
  pattern = r"(?:\+91|0)[6-9]\d{9}\b"
  return re.findall(pattern, text)

sentence = "Call me at +919990001234 or 09990004567"
print(extract_number(sentence))
```

```python
def extract_propernoun(text):
  pattern = r"(?<!^)\b[A-Z][a-z]+\b"
  return re.findall(pattern,text)

sentence = "Delhi is the capital of India. Narendra Modi is the Prime Minister."
print(extract_propernoun(sentence))
```
